```python
# Task 3 — Data Quality
#For each state, sum district-level transactions, amount, and registered users;
#compare these with the state-level values and identify discrepancies.

district_state_summary = (
    dist_txn_user.groupby("State")
    .agg(
        district_transactions = ("Transactions","sum"),
        district_amount = ("Amount (INR)","sum"),
        district_registered_users = ("Registered Users","sum")
    ).reset_index()
)
district_state_summary
```

```python
state_summary = (
    state_txn_users.groupby("State")
    .agg(
        state_transactions = ("Transactions","sum"),
         state_amount = ("Amount (INR)","sum"),
         state_registered_users = ("Registered Users","sum")
    ).reset_index()
)
state_summary
```

```python
comparision = pd.merge(
    state_summary,
    district_state_summary,
    on = "State",
    how = "inner"
)
comparision["transaction_difference"] = (
    comparision["state_transactions"]-comparision["district_transactions"]
)

comparision["Amount_difference"] = (
    comparision["state_amount"] - comparision["district_amount"]
)

comparision["Registered_users_diferent"] = (
    comparision["state_registered_users"] - comparision["district_registered_users"]
)

discrepancies = comparision[
    (comparision["transaction_difference"] != 0) | 
    (comparision["Amount_difference"] != 0) | 
    (comparision["Registered_users_diferent"] !=0 )
    ]

discrepancies
```

```python
# Task 4.1 — Registered Users / Population Ratio by State
# The question is:
# Merge state transaction/user data with district demographic data and calculate the registered 
# users-to-population ratio by state, then visualize it.

#Step 1 — Calculate total population by state
state_population = (
        district_demographics
        .groupby("State")["Population"]
        .sum()
        .reset_index()
    )
state_population
```

```python
# Step 2 — Calculate registered users by state

state_users = (
    state_txn_users.
    groupby("State")["Registered Users"]
    .sum()
    .reset_index()
)
state_users
```

```python
# Step 3 — Merge the two
users_population =pd.merge(
    state_users,
    state_population,
    on="State",
    how="inner"
)
users_population
```

```python
# ratio calculation
users_population["users_population_ratio"] = (
    users_population["Registered Users"]/users_population["Population"]
)
users_population["users_population_percentage"] = (
    users_population["Registered Users"]/users_population["Population"]
) * 100
users_population
```

```python
import seaborn as sns
import matplotlib.pyplot as plt

plt.figure(figsize = (15,7))
sns.barplot(
    data = users_population,
    x="State", y="users_population_percentage"
)
plt.xlabel("State")
plt.ylabel("users_population_percentage")
plt.title("Registered Users to Population Ratio by State")

plt.xticks(rotation = 90)
plt.tight_layout()
plt.show
```

```python
# Step 1 — Select the required columns

district_transactions = dist_txn_user[
    ["State","District","Transactions"]
]
district_desity = district_demographics[
    ["State","District","Density"]
]
```

```python
# Task 4.3 — Average Transaction Amount per User by State
# The goal is to calculate the average transaction amount per registered user for each state,
# then identify the top 5 and bottom 5 states.

# Step 1: Aggregate amount and registered users by state
state_amount_users = (
    state_txn_users
    .groupby("State")
    .agg(
        tottal_amount = ("Amount (INR)", "sum"),
        tottal_registered_users = ("Registered Users","sum")
    ).reset_index()
)
```
